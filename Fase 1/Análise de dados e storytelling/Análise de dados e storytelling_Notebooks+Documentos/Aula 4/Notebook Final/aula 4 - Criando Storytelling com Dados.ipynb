### Dengue (2025)

A dengue faz parte de um grupo de doenças denominadas arboviroses, que se caracterizam por serem causadas por vírus transmitidos por vetores artrópodes. No Brasil, o vetor da dengue é a fêmea do mosquito Aedes aegypti (significa "odioso do Egito"). Os vírus dengue (DENV) estão classificados cientificamente na família Flaviviridae e no gênero Flavivirus. Até o momento são conhecidos quatro sorotipos – DENV-1, DENV-2, DENV-3 e DENV-4 –, que apresentam distintos materiais genéticos (genótipos) e linhagens.<br/>

As evidências apontam que o mosquito tenha vindo nos navios que partiam da África com escravos. No Brasil, a primeira epidemia documentada clínica e laboratorialmente ocorreu em 1981-1982, em Boa Vista (RR), causada pelos sorotipos 1 e 4. Após quatro anos, em 1986, ocorreram epidemias atingindo o estado do Rio de Janeiro e algumas capitais da região Nordeste. Desde então, a dengue vem ocorrendo de forma continuada (endêmica), intercalando-se com a ocorrência de epidemias, geralmente associadas à introdução de novos sorotipos em áreas indenes (sem transmissão) e/ou alteração do sorotipo predominante, acompanhando a expansão do mosquito vetor.<br/>

Aspectos como a urbanização, o crescimento desordenado da população, o saneamento básico deficitário e os fatores climáticos mantêm as condições favoráveis para a presença do vetor, com reflexos na dinâmica de transmissão desses arbovírus. A dengue possui padrão sazonal, com aumento do número de casos e o risco para epidemias, principalmente entre os meses de outubro de um ano a maio do ano seguinte.

### Sinais e sintomas
A dengue é uma doença febril aguda, sistêmica, dinâmica, debilitante e autolimitada. A maioria dos doentes se recupera, porém, parte deles podem progredir para formas graves, inclusive virem a óbito. A quase totalidade dos óbitos por dengue é evitável e depende, na maioria das vezes, da qualidade da assistência prestada e organização da rede de serviços de saúde.

Todo indivíduo que apresentar febre (39°C a 40°C) de início repentino e apresentar pelo menos duas das seguintes manifestações - dor de cabeça, prostração, dores musculares e/ou articulares e dor atrás dos olhos – deve procurar imediatamente um serviço de saúde, a fim de obter tratamento oportuno.

Calcular a taxa de infecção por 100 mil habitantes

```python
df_dengue.head()
```

```python
infeccoes_estado
```

```python
df_populacao_2013_2023 = pd.read_excel('Populacao_2013_2023.xlsx')
df_populacao_2013_2023.head()
```

```python
df_populacao_2013_2023.rename(columns={"Município": "municipio"}, inplace=True)
```

```python
# Taxa de infecção por município e ano

# Garantir que as colunas-chave estejam padronizadas
df_dengue["ano"] = pd.to_datetime(df_dengue["data_infeccoes"]).dt.year
df_populacao_2013_2023.rename(columns={"Município": "municipio"}, inplace=True)

# Consolidar os casos por município, UF e ano
casos_municipio_ano = (
    df_dengue
    .groupby(["municipio", "uf", "ano"], as_index=False)["qtd_infeccoes"]
    .sum()
)

# Transformar a população de colunas anuais para linhas
populacao_municipio_ano = df_populacao_2013_2023.melt(
    id_vars=["municipio", "uf"],
    var_name="ano",
    value_name="populacao"
)
populacao_municipio_ano["ano"] = (
    populacao_municipio_ano["ano"].str.extract(r"(\d{4})")[0].astype(int)
)

# Unir casos e população e calcular a taxa por 100 mil habitantes
taxa_municipio_ano = casos_municipio_ano.merge(
    populacao_municipio_ano,
    on=["municipio", "uf", "ano"],
    how="left",
    validate="many_to_one"
)
taxa_municipio_ano["taxa_infeccao_100mil"] = (
    taxa_municipio_ano["qtd_infeccoes"] / taxa_municipio_ano["populacao"] * 100000
).round(2)

taxa_municipio_ano.sort_values(
    ["ano", "taxa_infeccao_100mil"], ascending=[True, False]
).head()
```

```python
# Gráfico da evolução da taxa de infecção por município

# Selecionar os 10 municípios com mais casos acumulados no período
top_municipios = (
    taxa_municipio_ano
    .groupby(["municipio", "uf"], as_index=False)["qtd_infeccoes"]
    .sum()
    .sort_values("qtd_infeccoes", ascending=False)
    .head(10)
)

# Manter somente esses municípios para uma visualização legível
taxa_top_municipios = taxa_municipio_ano.merge(
    top_municipios[["municipio", "uf"]],
    on=["municipio", "uf"],
    how="inner"
).sort_values(["ano", "taxa_infeccao_100mil"])

plt.figure(figsize=(14, 7))
for (municipio, uf), dados_municipio in taxa_top_municipios.groupby(["municipio", "uf"]):
    plt.plot(
        dados_municipio["ano"],
        dados_municipio["taxa_infeccao_100mil"],
        marker="o",
        linewidth=2,
        label=f"{municipio} ({uf})"
    )

plt.title("Evolução da taxa de infecção por dengue nos 10 municípios com mais casos")
plt.xlabel("Ano")
plt.ylabel("Taxa por 100 mil habitantes")
plt.xticks(sorted(taxa_top_municipios["ano"].unique()))
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.legend(title="Município", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
# Resumo dos principais resultados

resumo_por_ano = (
    taxa_municipio_ano
    .groupby("ano", as_index=False)
    .agg(
        casos_confirmados=("qtd_infeccoes", "sum"),
        municipios_analisados=("municipio", "nunique"),
        taxa_media=("taxa_infeccao_100mil", "mean")
    )
)

ano_maior_numero_casos = resumo_por_ano.loc[
    resumo_por_ano["casos_confirmados"].idxmax(), "ano"
]
municipio_maior_taxa = taxa_municipio_ano.loc[
    taxa_municipio_ano["taxa_infeccao_100mil"].idxmax()
]
municipios_maior_total = (
    taxa_municipio_ano
    .groupby(["municipio", "uf"], as_index=False)["qtd_infeccoes"]
    .sum()
    .sort_values("qtd_infeccoes", ascending=False)
    .head(10)
)

print(f"Ano com maior número de casos: {ano_maior_numero_casos}")
print(
    "Maior taxa observada: "
    f"{municipio_maior_taxa['municipio']} ({municipio_maior_taxa['uf']}) - "
    f"{municipio_maior_taxa['ano']}: "
    f"{municipio_maior_taxa['taxa_infeccao_100mil']:.2f} casos por 100 mil habitantes"
)
print("\nMunicípios com maior número total de casos no período:")
display(municipios_maior_total)
```

```python
# Comparação entre casos acumulados e taxa média

comparacao_municipios = municipios_maior_total.merge(
    taxa_municipio_ano
    .groupby(["municipio", "uf"], as_index=False)["taxa_infeccao_100mil"]
    .mean()
    .rename(columns={"taxa_infeccao_100mil": "taxa_media_100mil"}),
    on=["municipio", "uf"],
    how="left"
)
comparacao_municipios["rotulo"] = (
    comparacao_municipios["municipio"] + " (" + comparacao_municipios["uf"] + ")"
)

fig, eixos = plt.subplots(1, 2, figsize=(16, 7))

casos_plot = comparacao_municipios.sort_values("qtd_infeccoes")
eixos[0].barh(casos_plot["rotulo"], casos_plot["qtd_infeccoes"], color="#2f6690")
eixos[0].set_title("Casos confirmados acumulados")
eixos[0].set_xlabel("Quantidade de casos")
eixos[0].grid(axis="x", linestyle="--", alpha=0.4)

taxa_plot = comparacao_municipios.sort_values("taxa_media_100mil")
eixos[1].barh(taxa_plot["rotulo"], taxa_plot["taxa_media_100mil"], color="#d1495b")
eixos[1].set_title("Taxa média de infecção")
eixos[1].set_xlabel("Casos por 100 mil habitantes")
eixos[1].grid(axis="x", linestyle="--", alpha=0.4)

fig.suptitle("Casos absolutos e taxa média nos municípios com maior número de casos", fontsize=14)
plt.tight_layout()
plt.show()
```

## Conclusões e recomendações

- **2023 apresentou o maior volume de casos confirmados** no período analisado, indicando um ano que merece atenção especial no planejamento das ações de vigilância.
- Em números absolutos, **Brasília, São Paulo e Belo Horizonte** aparecem entre os municípios com maior quantidade acumulada de casos. Esse resultado deve ser interpretado junto ao tamanho da população.
- A taxa por 100 mil habitantes permite identificar municípios menores que podem não aparecer no ranking de casos absolutos. Na base analisada, **Santa Maria (RS), em 2023, apresentou a maior taxa observada**.
- Valores muito elevados de taxa devem ser validados antes da interpretação final. Recomenda-se conferir o denominador populacional, possíveis duplicidades, a cobertura de notificação e a compatibilidade entre município, UF e ano.
- Para orientar políticas públicas, recomenda-se combinar o ranking de casos absolutos com o ranking de taxas, monitorar a evolução anual e priorizar ações de prevenção, controle do vetor e comunicação de risco nos municípios que apresentam persistência ou aumento da taxa.

A análise descreve associações e padrões observados na base; ela não permite afirmar, sozinha, que um fator específico tenha causado os surtos.
